# AFM Gold Nanoparticle Height Analysis

> **Pipeline:** classical detection (Otsu, LoG) vs. deep-learning detection (YOLO),  
> followed by instance segmentation with SAM2 and quantitative height measurement.

This notebook implements a full analysis pipeline for Atomic Force Microscopy (AFM) images of gold nanoparticles:
1. Three particle **detection** methods are compared side-by-side.
2. Detected bounding boxes / blobs are refined into precise **instance masks** using SAM2.
3. Per-particle **heights** are extracted with ring-baseline correction.

## 3. Particle Detection — Method Comparison

Three approaches are evaluated on the same image:

| Method | Type | Key parameter |
|--------|------|---------------|
| **Otsu thresholding** | Classical | Global intensity threshold |
| **Laplacian of Gaussian (LoG)** | Classical, scale-space | Blob scale σ |
| **YOLO** | Deep learning | Confidence threshold |

The section shows detection maps and particle counts for each method.

### 3.1 Otsu Thresholding (Classical Baseline)

Global threshold computed by Otsu's method on `z_result`. Connected components above the threshold are treated as individual particles.  
Simple and fast, but sensitive to background variation and particle merging.

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# ── Панель 1: z_above ─────────────────────────────────────────
ax = axes[0]
im = ax.imshow(z_result, cmap="afmhot", origin="upper")
plt.colorbar(im, ax=ax, label="Z, нм", fraction=0.046, pad=0.04)
ax.set_title("z_above (частицы над подложкой)")
ax.axis("off")

# ── Панель 2: бинарная маска ──────────────────────────────────
thresh = threshold_otsu(z_result)
binary = z_result > thresh

ax = axes[1]
ax.imshow(binary, cmap="gray", origin="upper")
ax.set_title(f"Otsu бинаризация (порог = {thresh:.2f} нм)")
ax.axis("off")

# ── Панель 3: размеченные объекты ─────────────────────────────
labeled = label(binary)
props   = regionprops(labeled)
radii_px = np.array([p.equivalent_diameter_area / 2 for p in props])

ax = axes[2]
ax.imshow(z_result, cmap="afmhot", origin="upper", alpha=0.6)
for p in props:
    y, x    = p.centroid
    r       = p.equivalent_diameter_area / 2
    circle  = plt.Circle(
        (x, y), r,
        color="cyan", fill=False, linewidth=1.0, alpha=0.8
    )
    ax.add_patch(circle)
ax.set_title(f"Объекты Otsu: {len(props)}  |  "
             f"радиусы {radii_px.min():.1f}–{radii_px.max():.1f} пкс")
ax.axis("off")

plt.suptitle(
    f"Otsu бинаризация z_above  |  порог = {thresh:.2f} нм",
    fontsize=12
)
plt.tight_layout()
plt.show()

# Статистика
print(f"Порог Otsu:        {thresh:.3f} нм")
print(f"Объектов найдено:  {len(props)}")
print(f"Радиусы (пкс):     {radii_px.min():.1f} – {radii_px.max():.1f}")
print(f"Медиана радиуса:   {np.median(radii_px):.1f} пкс  "
      f"= {np.median(radii_px) * pixel_size_nm:.1f} нм")

radii = sizes["radii_px"]
print(f"radii_px: min={radii.min():.2f}  max={radii.max():.2f}  "
      f"median={np.median(radii):.2f}  n={len(radii)}")

plt.figure(figsize=(7, 3))
plt.hist(radii, bins=30, color="steelblue", edgecolor="white")
plt.xlabel("Радиус, пкс")
plt.ylabel("Количество")
plt.title("Распределение радиусов Otsu")
plt.axvline(np.median(radii), color="gold", linestyle="--",
            label=f"Медиана: {np.median(radii):.1f} пкс")
plt.legend()
plt.tight_layout()
plt.show()
```

## 4. Instance Segmentation with SAM2

Both LoG blob centres and YOLO bounding boxes are used as prompts for SAM2, which produces a precise binary mask for each detected particle.  
Masks enable accurate **ring-baseline** height extraction: substrate level is estimated from a dilated annulus around each mask.

```python
import os
os.environ["CUDA_VISIBLE_DEVICES"] = ""

import torch
from sam2.build_sam import build_sam2
from sam2.sam2_image_predictor import SAM2ImagePredictor


sam2 = build_sam2(
    config_file="configs/sam2.1/sam2.1_hiera_b+.yaml",
    ckpt_path="checkpoints/sam2.1_hiera_base_plus.pt",
    device='cpu',
)

predictor = SAM2ImagePredictor(sam2)
print("SAM загружен")
```

### 4.1 LoG → SAM2

LoG blob centres (y, x) and estimated radii are passed as point + box prompts.  SAM2 returns the best-scoring mask out of three candidates.

```python
df_sam2, sam2_masks = run_sam2_from_blobs(
    predictor, z_flat, z_result, blobs, pixel_size_nm
)

print(f"{len(df_sam2)} частиц измерено")
print(df_sam2.height_nm.describe().round(2))
print(f"Масок сохранено: {len(sam2_masks)}")
```

```python
overlay = np.zeros((*z_result.shape, 4))  # RGBA

for m in sam2_masks:
    mask = m['mask']
    
    overlay[mask, 0] = 0.0   # R
    overlay[mask, 1] = 1.0  # G
    overlay[mask, 2] = 1   # B  → голубой
    overlay[mask, 3] = 1   # alpha (прозрачность)

fig, ax = plt.subplots(1, 3, figsize=(14, 5))

ax[0].imshow(z_result, cmap='afmhot', origin='lower')
ax[0].imshow(overlay, alpha=0.6, origin='lower')

for m in sam2_masks:
    ax[0].scatter(m['x_px'], m['y_px'], c='yellow', s=10)


h, w = z_result.shape
ticks_px = np.linspace(0, w - 1, 5)
ticks_py = np.linspace(0, h - 1, 5)
ax[0].set_xticks(ticks_px)
ax[0].set_yticks(ticks_py)
ax[0].set_xticklabels([f"{v * pixel_size_nm:.0f}" for v in ticks_px])
ax[0].set_yticklabels([f"{v * pixel_size_nm:.0f}" for v in ticks_py])
ax[0].set_xlabel("X, nm")
ax[0].set_ylabel("Y, nm")
ax[0].set_title(f"SAM2 Masks + centers: {len(sam2_masks)} particles")

plot_detections(z_result, blobs, pixel_size_nm, ax[1])
ax[2].imshow(z_result, cmap='afmhot', origin='lower')
ax[2].set_title('Initial topology')
ax[2].set_xticks(ticks_px)
ax[2].set_yticks(ticks_py)
ax[2].set_xticklabels([f"{v * pixel_size_nm:.0f}" for v in ticks_px])
ax[2].set_yticklabels([f"{v * pixel_size_nm:.0f}" for v in ticks_py])
ax[2].set_xlabel("X, nm")
ax[2].set_ylabel("Y, nm")

plt.savefig('images/log.png', dpi=300)
plt.show()
```

## 5. Height Measurement

For each particle mask the height is defined as:

$$h = \text{max}(Z_{\text{inner mask}}) - \text{median}(Z_{\text{ring}})$$

where the **inner mask** is eroded by a few pixels to avoid edge artefacts, and the **ring** is a dilated annulus restricted to the substrate region.

### 5.1 Baseline Correction — Single-Particle Profile

Visual verification of mask, ring, and Z profile for one example particle.

```python
row = measurements.iloc[0]
y_i, x_i = int(row.y_px), int(row.x_px)
radius_px = row.sigma_px * np.sqrt(2)

mask       = create_circular_mask(z_flat.shape, y_i, x_i, radius_px)
otsu_thresh    = threshold_otsu(z_flat)
substrate_mask = z_flat < otsu_thresh
clean_ring = get_clean_ring(mask, substrate_mask, outer_px=outer_px, inner_erode_px=inner_px)

# --- правильный подсчёт ---
mask_vals  = z_flat[mask]
ring_vals  = z_flat[clean_ring]
baseline   = np.median(ring_vals)          # медиана кольца — устойчива к выбросам
peak       = np.max(mask_vals)  # p99 а не max — защита от горячих пикселей
height     = peak - baseline

# --- визуализация ---
half = 20
profile_crop = z_flat[y_i, x_i-half : x_i+half]
coords       = (np.arange(2*half) - half) * pixel_size_nm

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Маска и кольцо
overlay = np.zeros(z_flat.shape, dtype=float)
overlay[mask]       = 1
overlay[clean_ring] = 2
axes[0].imshow(z_flat[y_i-half:y_i+half, x_i-half:x_i+half], cmap='afmhot')
axes[0].imshow(overlay[y_i-half:y_i+half, x_i-half:x_i+half], alpha=0.4, cmap='cool')
axes[0].set_title(f'Mask and ring | outer={outer_px}, inner_erode={inner_px}')

# Профиль
axes[1].plot(coords, profile_crop, color='gold')
axes[1].axhline(baseline, color='cyan',  linestyle='--',
                label=f'baseline (median ring) = {baseline:.2f} nm')
axes[1].axhline(peak,     color='red',   linestyle='--',
                label=f'peak (p99 mask) = {peak:.2f} nm')
axes[1].fill_between(coords, baseline, peak,
                     alpha=0.1, color='red')
axes[1].axvline(0, color='white', linestyle=':', alpha=0.4, label='center')
axes[1].set_xlabel('Δ от центра, nm')
axes[1].set_ylabel('Z, nm')
axes[1].set_title(f'Height = {height:.2f} nm  |  SNR = {height/ring_vals.std():.1f}')
axes[1].legend(fontsize=9)
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig('images/example_measurement.png', dpi=300)
plt.show()

print(f'baseline  = {baseline:.3f} nm  (median ring, n={clean_ring.sum()} px)')
print(f'peak      = {peak:.3f} nm  (p99 mask,  n={mask.sum()} px)')
print(f'height    = {height:.3f} nm')
print(f'ring std  = {ring_vals.std():.3f} nm')
print(f'SNR       = {height/ring_vals.std():.1f}')
```

## 6. Results and Comparison

Quantitative comparison of all three pipelines on the same AFM image:

| Pipeline | Detection | Segmentation | Variable |
|----------|-----------|--------------|----------|
| **LoG + Baseline** | LoG blobs | circular ring mask | `measurements` |
| **LoG + SAM2** | LoG blobs | SAM2 instance mask | `df_sam2` |
| **YOLO + SAM2** | YOLO bboxes | SAM2 instance mask | `df_yolo` |

The overlay histogram and box plot reveal differences in particle count, height median, and distribution shape.  
A combined long-format DataFrame `df_all` is exported for downstream analysis.

```python
from scipy.stats import ks_2samp

palette = {
    'LoG + Baseline': 'steelblue',
    'LoG + SAM2':     'darkorange',
    'YOLO + SAM2':    'seagreen',
}
series = {
    'LoG + Baseline': measurements.height_nm.dropna(),
    'LoG + SAM2':     df_sam2.height_nm.dropna(),
    'YOLO + SAM2':    df_yolo.height_nm.dropna(),
}

# ── Summary statistics ────────────────────────────────────────────────────────
summary = pd.DataFrame({k: v.describe() for k, v in series.items()}).round(2)
print("=== Height statistics (nm) ===")
print(summary)

# ── KS-test: algo vs neural ───────────────────────────────────────────────────
for label_a, label_b in [('LoG + Baseline', 'YOLO + SAM2'),
                          ('LoG + SAM2',     'YOLO + SAM2')]:
    stat, p = ks_2samp(series[label_a], series[label_b])
    print(f"\nKS({label_a} vs {label_b}):  D={stat:.3f},  p={p:.4f}")

# ── Overlay histograms ────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

ax = axes[0]
for label, data in series.items():
    ax.hist(data, bins=40, alpha=0.5, color=palette[label],
            edgecolor='white', label=f'{label}  (n={len(data)})')
    ax.axvline(data.median(), color=palette[label],
               linestyle='--', linewidth=1.4, alpha=0.9)
ax.set_xlabel('Height, nm')
ax.set_ylabel('Count')
ax.set_title('Height distributions — overlay\n(dashed = median)')
ax.legend(fontsize=9)
ax.grid(alpha=0.3)

# ── Box plots ─────────────────────────────────────────────────────────────────
ax = axes[1]
labels_bp = [l.replace(' + ', '\n+ ') for l in series]
bp = ax.boxplot(list(series.values()), labels=labels_bp,
                patch_artist=True, notch=False,
                medianprops=dict(color='white', linewidth=2))
for patch, color in zip(bp['boxes'], palette.values()):
    patch.set_facecolor(color)
    patch.set_alpha(0.75)
ax.set_ylabel('Height, nm')
ax.set_title('Height distributions — box plot')
ax.grid(alpha=0.3, axis='y')

plt.suptitle('Method comparison — height measurement', fontsize=13)
plt.tight_layout()
plt.savefig('images/results.png', dpi=300)
plt.show()

# ── Long-format combined DataFrame ───────────────────────────────────────────
# measurements не хранит peak_nm, вычисляем: peak = height + baseline
m_base = measurements[['x_px', 'y_px', 'height_nm', 'baseline_nm']].copy()
m_base['peak_nm'] = m_base['height_nm'] + m_base['baseline_nm']
m_base['pipeline'] = 'LoG + Baseline'

m_sam2 = df_sam2[['x_px', 'y_px', 'height_nm', 'baseline_nm', 'peak_nm']].copy()
m_sam2['pipeline'] = 'LoG + SAM2'

m_yolo = df_yolo[['x_px', 'y_px', 'height_nm', 'baseline_nm', 'peak_nm']].copy()
m_yolo['pipeline'] = 'YOLO + SAM2'

df_all = pd.concat([m_base, m_sam2, m_yolo], ignore_index=True)
print(f"\ndf_all shape: {df_all.shape}")
df_all.groupby('pipeline')['height_nm'].agg(['count', 'median', 'mean', 'std']).round(2)
```
